# Week 3 pair-programming exercises

## Introduction to pair-programming exercises

You will work in pairs (or, in rare case, groups of 3) on this set of exercises together, submitting a single notebook file at the end of the class period. Your partner(s) will rotate every 2 weeks.

### Workflow

Each question will be timed to ensure that everyone gets to work on at least a part of every question part. When working on the exercises, one member of the pair/group should serve as the **"driver"**, while the remaining member(s) should serve as **"navigator(s)."** The responsibilities of these roles are as follows:

- **Driver**: the driver is responsible for the detailed implementations of the coding solution, and should be the sole person typing out the actual codes. As the driver types out codes, the driver should talk aloud about one's thought processes so that these can be discussed.
  
- **Navigator**: the navigator is responsible for navigating the bigger picture of the problem. The navigator should provide feedback to the driver about the codes that have already been written (e.g., raise questions if the purpose of a segment of codes is unclear), and help debug any issues that arise. In addition, the navigator is responsibile for making suggestions on how to proceed, and for searching through resources (e.g., official documentation) that may assist the driver in completing the task.

The roles of driver and navigator(s) should be rotated for **every subpart of the question**. To facilitate rapid switching, a single computer should be used by all drivers and the remaining computers should be used by the navigator(s).

### Use of AI / online resources

- The **driver** should refrain from using generative AI or from navigating to the broader internet beyond the JupyterHub.

- The **navigator** should refrain from using generative AI, but should feel free to consult internet resources (e.g., official documentations) in assisting the driver. However, the navigator should make sure that they are relaying information **in their own words** when assisting the driver.

### Best practices in coding

You should try to follow the best practices in coding (which you will be graded on in homework and final project). The best practices include:

1) Variables are used to store objects, and their names are meaningful.

2) Codes are formatted consistently

4) Comments are used to document your intention, as well as (less commonly) tricky implementation

5) Help from outside sources (e.g., other groups or online documentations) are documented. 

6) Final notebook fully runs from start to finish (a good way to check this is restarting the kernel and running through all the cells).
